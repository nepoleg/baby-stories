```python
from transformers import AutoModelForCausalLM, PreTrainedTokenizerFast
```

```python
path = "./custom-gpt"

tokenizer = PreTrainedTokenizerFast.from_pretrained(path)
model = AutoModelForCausalLM.from_pretrained(path, trust_remote_code=True) #to(device)
```

```python
inputs = tokenizer("<|endoftext|>", return_tensors="pt") #.to(device)
outputs = model.generate(**inputs, max_new_tokens=500, use_cache=False, do_sample=True, num_beams=4, num_return_sequences=4)
res = tokenizer.batch_decode(outputs, skip_special_tokens=True)
print(f"{res=}")
```
